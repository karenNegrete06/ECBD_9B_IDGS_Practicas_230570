22. Calculando los datos para graficación total de estadisticas por cada tipo de pokemon, respetando el orden alfabetico previamente definido

```python
plt.style.use('seaborn-v0_8-darkgrid')

stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']

palettes = ['magma', 'ocean', 'vlag', 'copper', 'mako', 'winter']

# Agrupar una sola vez
pokemon_stats = pokemons.groupby('Type')[stats].sum()

fig, axes = plt.subplots(3, 2, figsize=(17, 17))

for ax, stat, palette in zip(axes.flat, stats, palettes):

    sns.barplot(
        data=pokemon_stats.reset_index(),
        x=stat,
        y='Type',
        hue='Type',
        palette=palette,
        dodge=False,
        ax=ax
    )

    # Eliminar leyenda
    if ax.legend_:
        ax.legend_.remove()

    # Etiquetas de valores
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt='%.0f',
            padding=3
        )

    ax.set_title(f'Total de {stat}')
    ax.set_xlabel(stat)
    ax.set_ylabel('Tipo')

plt.tight_layout()
plt.show()
```

24. Graficación de estadistica por tipo

```python
import math

k = 1
n = len(list_types)
cols = 3
rows = math.ceil(n / cols)

fig, axes = plt.subplots(rows, cols, figsize=(17, rows * 4))
axes = axes.flatten()

for ax, tipo in zip(axes, list_types):

    data_type = pokemons[pokemons['Type'] == tipo]

    totals = (
        data_type
        .sum(numeric_only=True)
        .drop(['#', 'Generation', 'Total'], errors='ignore')
    )

    df_totals = totals.reset_index()
    df_totals.columns = ['Stat', 'Value']

    sns.barplot(
        data=df_totals,
        x='Value',
        y='Stat',
        hue='Stat',
        palette='inferno',
        dodge=False,
        ax=ax
    )

    # Quitar leyenda
    if ax.legend_:
        ax.legend_.remove()

    # Etiquetas de valores
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt='%.0f',
            padding=3
        )

    ax.set_title(tipo)
    ax.set_xlabel('Total')
    ax.set_ylabel('Estadística')

# Ocultar subplots vacíos
for ax in axes[len(list_types):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

**Ejercicio 15:** Obtener el conjunto de graficos por cada estadistica por tipo de pokemon en el DF modificado

```python
import math

list_types = pokemonsEvo['Type'].unique()

k = 1
n = len(list_types)
cols = 3
rows = math.ceil(n / cols)

plt.figure(figsize=(17, rows * 4))

for i in list_types:
    plt.subplot(rows, cols, k)
    k += 1

    data_type = pokemonsEvo[pokemonsEvo['Type'] == i]

    totals = data_type.select_dtypes(include='number').sum()

    totals = totals.drop(
        labels=['#', 'Generation', 'Total'],
        errors='ignore'
    )

    df_totals = totals.reset_index()
    df_totals.columns = ['Stat', 'Value']

    ax = sns.barplot(
        data=df_totals,
        x='Value',
        y='Stat',
        hue='Stat',
        palette='inferno',
        dodge=False
    )

    # Eliminar leyenda
    if ax.legend_:
        ax.legend_.remove()

    # Mostrar valores al final de las barras
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt='%.0f',
            padding=3
        )

    plt.title(i)
    plt.xlabel('Total')
    plt.ylabel('Estadística')

plt.tight_layout()
plt.show()
```

26. Utilizar graficos e enjambre (SwarmPlota) para representar los valores especificos de las estadisticas por el tipo de  Pokemon

```python
list_types= pokemons['Type'].unique().tolist()
list_types.sort()

n_cols = 3
n_rows = math.ceil(len(list_types) / n_cols)

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(17, 4 * n_rows),
    sharey=True
)

axes = axes.flatten()

for ax, tipo in zip(axes, list_types):

    data_tipo = pok_melt[pok_melt['Type'] == tipo]

    sns.stripplot(
        data=data_tipo,
        x='variable',
        y='value',
        hue='variable',
        palette='gist_stern',
        jitter=True,
        size=4,
        alpha=0.7,
        legend=False,
        ax=ax
    )

    ax.set_title(tipo)
    ax.set_xlabel("")
    ax.set_ylabel("Valor")
    ax.tick_params(axis='x', rotation=45)


for ax in axes[len(list_types):]:
    ax.remove()

plt.tight_layout()
plt.show()
```
