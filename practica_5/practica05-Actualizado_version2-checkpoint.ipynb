15. Conteo de los pokemones por el atributo tipo (Graficado) se usa la libreria **matplot** para configurar el tamaño de la grafica 

**NOTA:** Por el número de pokemones de tipo agua y normal es alto comparado al resto , los pokemones respectivos tendran siempre unn total de stats as altos 

```python
plt.figure(figsize=(10, 5))

ax = sns.countplot(
    data=pokemons,
    x='Type',
    hue='Type',
    palette='twilight',
    legend=False
)

for container in ax.containers:
    ax.bar_label(
        container,
        label_type='edge',
        padding=3
    )

plt.title('Número de Pokémon agrupados por tipo')
plt.xlabel('Tipo')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Ejercicio 09:** Obtener el grafico de total de pokemones por tipo en el DF modificado. Ordenados alfabeticamente

```python
orden = sorted(pokemons['Type'].unique())

ax = sns.countplot(
    data=pokemons,
    x='Type',
    order=orden,
    hue='Type',
    palette='twilight',
    legend=False
)

# Agregar totales sobre cada barra
for container in ax.containers:
    ax.bar_label(
        container,
        label_type='edge',
        padding=3
    )

plt.title('Total de Pokémones por Tipo (Ordenados Alfabéticamente)')
plt.xlabel('Tipo')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

22. Calculando los datos para graficación total de estadisticas por cada tipo de pokemon, respetando el orden alfabetico previamente definido

```python
plt.style.use('seaborn-v0_8-darkgrid')    # Plantilla de colores

stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']    # Arreglo de Estadísticas

palettes = ['magma', 'ocean', 'vlag', 'copper', 'mako', 'winter']     # Paleta de colores para cada gráfico de cada estadística


# Agrupar una sola vez
pokemon_stats = pokemons.groupby('Type')[stats].sum()

plt.figure(figsize=(17, 17))

for k, stat in enumerate(stats, start=1):

    plt.subplot(3, 2, k)

    sns.barplot(
        x=pokemon_stats[stat],
        y=pokemon_stats.index,
        hue=pokemon_stats.index,    # evita warning de palette
        palette=palettes[k-1],
        legend=False
    )

    plt.title(f'Total of {stat}')
    plt.xlabel(stat)
    plt.ylabel('Type')

plt.tight_layout()
plt.show()
```
